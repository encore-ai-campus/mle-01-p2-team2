```python
import requests
import json

API_KEY = "1592d1645c2e470fa950"
SERVICE_ID = "COOKRCP01"

BASE_URL = "https://openapi.foodsafetykorea.go.kr/api"

all_recipes = []

start_idx = 1
batch_size = 1000

while True:
    end_idx = start_idx + batch_size - 1

    url = (
        f"{BASE_URL}/{API_KEY}/{SERVICE_ID}/json/"
        f"{start_idx}/{end_idx}"
    )

    response = requests.get(url)
    response.raise_for_status()

    data = response.json()

    result = data[SERVICE_ID]

    # 전체 데이터 개수
    total_count = int(result["total_count"])

    # 현재 가져온 데이터
    rows = result.get("row", [])

    all_recipes.extend(rows)

    print(
        f"{start_idx} ~ {end_idx} 수집 완료 "
        f"({len(all_recipes)} / {total_count})"
    )

    # 전체 데이터 수집 완료
    if len(all_recipes) >= total_count:
        break

    start_idx += batch_size


print("총 레시피 수:", len(all_recipes))
```

```python
with open(
    "mfds_recipes.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        all_recipes,
        f,
        ensure_ascii=False,
        indent=2
    )
```
