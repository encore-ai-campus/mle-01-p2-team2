```python
import pandas as pd

input_path = saved_path
clean_path = result_dir / "recipes_classified_cleaned.jsonl"

df = pd.read_json(input_path, lines=True)

print("원본:", len(df))


def has_value(value):
    if value is None:
        return False

    if isinstance(value, list):
        return len(value) > 0

    if isinstance(value, str):
        return bool(value.strip())

    if pd.isna(value):
        return False

    return True


mask = (
    (df["graph_eligible"] == True)
    & df["title"].apply(has_value)
    & df["cooking_method"].apply(has_value)
    & df["ingredients_clean"].apply(has_value)
)

df_clean = df[mask].copy()

print("전처리 후:", len(df_clean))
print("제거된 데이터:", len(df) - len(df_clean))

df_clean.to_json(
    clean_path,
    orient="records",
    lines=True,
    force_ascii=False
)

print("전처리 파일:", clean_path)

# 이후 10번 셀이 사용할 데이터로 다시 로드
classified_recipes = read_recipe_jsonl(clean_path)
```
