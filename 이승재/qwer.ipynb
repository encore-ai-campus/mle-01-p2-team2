```python
import json
from pathlib import Path

path = Path(r"C:\Users\Playdata\Desktop\mle-01-p2-team2\전처리\recipes_dish_final.jsonl")

dish_names = []

with path.open("r", encoding="utf-8") as f:
    for line in f:
        if not line.strip():
            continue
        data = json.loads(line)
        name = (data.get("dish") or {}).get("name")
        if name:
            dish_names.append(name)
```

```python

from collections import Counter

# 문자열과 딕셔너리가 섞인 값을 요리명 문자열로 통일
clean_names = []

for item in dish_names:
    while isinstance(item, dict):
        item = item.get("name")

    if isinstance(item, str) and item.strip():
        clean_names.append(item.strip())

# 요리명별 개수: 많은 순서로 출력
dish_counts = Counter(clean_names)

for name, count in dish_counts.most_common():
    print(f"{name}: {count}")
```

```python
import pandas as pd

dish_summary = (
    pd.Series(clean_names, name="dish")
    .value_counts()
    .rename_axis("dish")
    .reset_index(name="count")
)

with pd.option_context("display.max_rows", None):
    display(dish_summary)
```
